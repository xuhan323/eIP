## UDD test : Uncertainty-Driven Molecular Dynamics Simulations

This notebook provides guidance for running uncertainty-driven molecular dynamics (UDD) simulations. The UDD approach incorporates uncertainty quantification into molecular dynamics simulations, which allows for better sampling of conformational space and identification of regions with high prediction uncertainty.


### Key Features

- Uses trained interatomic potential models with uncertainty estimates
- Applies biasing forces based on uncertainty to enhance sampling
- Captures both aleatoric (data) and epistemic (model) uncertainty during simulation
- Produces trajectory files and energy/uncertainty logs for analysis

All conformations are derived from aspirin molecules.

## Output Files Description

The UDD-MD simulation produces two main types of output files:

1. Trajectory File: Contains atomic positions saved at regular intervals
2. Energy and Uncertainty Log: Records energy components and uncertainty values

### Trajectory File

The trajectory file is named according to the parameters used (e.g., A_4500_B10_C5_D200.traj) and contains atomic configurations saved every sampling steps. This file can be analyzed using tools like ASE (Atomic Simulation Environment) or visualized in software like VMD or OVITO.

### Energy and Uncertainty Log
The energy and uncertainty log file contains records of various energy components and uncertainty values throughout the simulation. This file is saved in the specified log directory and can be analyzed to understand the energy evolution and uncertainty changes during the simulation.
